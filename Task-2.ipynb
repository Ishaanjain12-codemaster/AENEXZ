```python
unit=int(input("Enter the number of units:- "))
bill=0

if(unit<=100):
    bill=100*5
elif(unit>100 and unit<=200):
    bill=(100*5)+((unit-100)*7)
elif(unit>200):
    bill=(100*5)+(100*7)+((unit-200)*10)

print("Final Bill Amount: ",bill)
```

```python
amt=float(input("Enter the amount:- "))
dis_amt=0
if(amt>=5000):
    print("Original Amount: ",amt)
    dis_amt=amt-(0.20*amt)
    print("Discounted Amount: ",dis_amt)
elif(amt>=2000 and amt<=4999):
    print("Original Amount: ",amt)
    dis_amt=amt-(0.10*amt)
    print("Discounted Amount: ",dis_amt)
else:
    print("Original Amount: ",amt)
    print("Discounted Amount: No Discount for such low amount")
```
