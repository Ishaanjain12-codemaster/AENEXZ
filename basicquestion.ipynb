```python
# Write a program to find the largest of three numbers.

def largest(num):
    max=0
    for i in num:
        if(i>max):
            max=i
    return max

a=int(input("Enter the number: ")) 
b=int(input("Enter the number: ")) 
c=int(input("Enter the number: ")) 
num=[a,b,c]
print(largest(num))
```

```python
# Write a function that takes the radius of a circle and returns its area.

import math

def area(r):
    area = math.pi * r * r
    return area

a=int(input("Enter the number: "))
print("Area of circle: ",area(a))
```

```python
# Write a function that accepts three numbers and returns the largest one.

def largest(a,b,c):
    num=[a,b,c]
    max=0
    for i in num:
        if(i>max):
            max=i
    return max

a=int(input("Enter the number: ")) 
b=int(input("Enter the number: ")) 
c=int(input("Enter the number: ")) 
print("Largest Number is ",largest(a,b,c))
```

```python
# Write a function that checks whether a number is prime.

def is_prime(n):
    for i in range(2,n):
        if n%i==0:
            return False
            
    return True

a=int(input("Enter the number: "))
if is_prime(a):
    print("Is Prime")
else:
    print("Not a Prime")
```

```python
# Write a recursive function to find the nth Fibonacci number.

def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    return fib(n - 1) + fib(n - 2)

print(fib(3))
```

```python
# Write a recursive function to reverse a string.

def reverse(s):
    if len(s) == 0:
        return s
    return reverse(s[1:]) + s[0]

text = input("Enter a string: ")
print("Reversed string:", reverse(text))
```

```python
# Write a program to find the sum of all numbers from 1 to n using a for loop.

def sum(n):
    s=0
    for i in range(1,n+1):
        s+=i
    return s

print(sum(10))
```

```python
# Write a program to count how many even numbers are present in a list.

n=int(input("Enter the number of integers in the list: "))
lst=[]
even=0
odd=0

for i in range(0,n):
    a=int(input("Enter the numbers"))
    lst.append(a)

for i in lst:
    if i%2==0:
        even+=1
    else:
        odd+=1
print("The List of integers:",lst)
print("Even Count:",even)
print("Odd Count:",odd)
```
